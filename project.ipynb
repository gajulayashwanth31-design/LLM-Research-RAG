```python
from collections import Counter

chunk_counts = Counter(
    chunk.metadata.get("source_file")
    for chunk in chunks
)

for file, count in chunk_counts.items():
    print(f"{file}: {count} chunks")
```

```python
for i, chunk in enumerate(chunks[:5]):
    print(f"\n{'='*80}")
    print(f"CHUNK {i}")
    print(f"Metadata: {chunk.metadata}")
    print(f"\nContent:\n{chunk.page_content[:1000]}")
```

```python
lengths = [len(chunk.page_content) for chunk in chunks]

print(f"Total chunks: {len(lengths)}")
print(f"Smallest: {min(lengths)} characters")
print(f"Largest: {max(lengths)} characters")
print(f"Average: {sum(lengths) / len(lengths):.2f} characters")
```
